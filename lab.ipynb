```python
import numpy as np 
import pandas as pd 
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],
    'Value': [10, 15, 20, 25, 30, 35,40,50],
    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],
    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000,500,600]
}
df = pd.DataFrame(data)
df
```

```python
# calculate total sum of the Value column
total_sum = df['Value'].aggregate('sum')
print("Total Sum:", total_sum)

# Alternate

total_sum2 = df[['Value']].sum()
print("Total Sum:", total_sum2)
```

```python
df[['Value']].cumsum()
```

```python
# calculate the mean of the Value column
mean1 = df['Value'].agg('mean')
print(f'Mean: {mean1}')

# Alternate

mean2 = df[['Value']].mean()
print(f'Mean: {mean2}')
```

```python
# calculate the maximum value in the Value column
max1 = df['Value'].agg('max')
print(f'Mean: {max1}')


max2 = df[['Value']].max()
print(f'Mean: {max2}')
```

```python
# Do both the agg and alternative to find: count
count1 = df['Value'].agg('count')
print(count1)

count2 = df[['Value']].count()
print(count2)
```

### Multiple Aggregations

```python
result = df [['Fee', 'Discount']].agg('sum')

result
```

```python
result = df [['Fee', 'Discount']].sum()

result
```

```python
result = df[['Fee', 'Discount']].agg(['min', 'max', 'mean'])

result
```

```python
result_group = df.groupby('Category')[['Value', 'Fee', 'Discount']].agg('sum')
print(result_group)
```

```python
result_group2 = df[['Fee', 'Discount']].agg(['min', 'max', 'mean'])

print(result_group2)
```
